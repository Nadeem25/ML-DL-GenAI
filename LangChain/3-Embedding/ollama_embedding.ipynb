```python
from langchain_community.embeddings import OllamaEmbeddings
embeddings = (
    OllamaEmbeddings(model='gemma:2b') # default
)
embeddings
```

```python
r1 = embeddings.embed_documents(
    [
        "Morgen muss ich universitate gehen",
        "Gestern musste ich buro gehen"
    ]
)
len(r1[0])
r1[0]
```
