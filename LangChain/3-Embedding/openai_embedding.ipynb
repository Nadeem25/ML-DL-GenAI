# Converting the text into vector

```python
import os
from dotenv import load_dotenv
load_dotenv() # Load all the environment variables
os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")
```

```python
from langchain_openai import OpenAIEmbeddings
text_embedding_3_large = OpenAIEmbeddings(model="text-embedding-3-large", dimensions=1024)
text_embedding_3_large

text = "This is a tutorial on OPENAI embedding"
query_result = text_embedding_3_large.embed_query(text)
query_result

len(query_result)
```

```python
# step 1. Read the text file and convert it into document
from langchain_community.document_loaders import TextLoader
loader= TextLoader('C:/Notes/ML-DL-GenAI/LangChain/Files/data.txt')
text_document = loader.load()
text_document

# Step 2. Vector embedding and vectore store in db
from langchain_community.vectorstores import Chroma
chroma_db = Chroma.from_documents(text_document, text_embedding_3_large)

# Step 3. Search in the vector db (Similarity Search)
query = "My name ist"
retrieved_result = chroma_db.similarity_search(query)
print(retrieved_result)
```
