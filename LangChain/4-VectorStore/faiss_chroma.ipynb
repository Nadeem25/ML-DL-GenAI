```python
# Facebook AI Similarity Search (FAISS) is a library for efficient similarity search and clustering of dense vectors. 
# It contains algorithms that search in sets of vectors of any size, up to ones that possibly do not fit in RAM. It also contains supporting code for evalution and parameter tuning.
```

```python
from langchain_community.document_loaders import TextLoader
from langchain_community.vectorstores import FAISS
from langchain_community.embeddings import OllamaEmbeddings
from langchain_text_splitters import CharacterTextSplitter


# Read the text file
loader= TextLoader('C:/Notes/ML-DL-GenAI/LangChain/Files/speech.txt')
# Convert the content into document
text_document = loader.load()
text_document
```

```python
text_splitter = CharacterTextSplitter(chunk_size = 200, chunk_overlap = 30)
char_docs = text_splitter.split_documents(text_document)
char_docs
```

```python
embeddings = (
    OllamaEmbeddings(model='gemma:2b') # default
)
db = FAISS.from_documents(char_docs, embeddings)
```

```python
# Query
query = "How does the speaker describe the desired outcome of the war?"
docs = db.similarity_search(query)
docs[0].page_content
```

```python
# As a Retriver: It is act like an interface
retriever = db.as_retriever()
docs = retriever.invoke(query)
docs[0].page_content
```

Similarity Search With Score

--> It is FAISS method. It allows you to return not only the documents but also the distance score of the query to them. The returned distance score is L2 distance. Therefore, a lowers score is better

```python
docs_with_score = db.similarity_search_with_score(query)
docs_with_score
```

Chroma: 

Is is AI-native open source vectore database focused on developer productivity and happiness. Chroma is licesed unded Apache 2.0

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.vectorstores import Chroma
text_splitter = RecursiveCharacterTextSplitter(chunk_size = 500, chunk_overlap = 0)
splits_docs = text_splitter.split_documents(text_document)
splits_docs
```

```python
embeddings = (
    OllamaEmbeddings(model='gemma:2b') # default
)
vectordb = Chroma.from_documents(documents=splits_docs, embedding=embeddings)
vectordb
```

```python
# Query
query = "How does the speaker describe the desired outcome of the war?"
docs = vectordb.similarity_search(query)
docs[0].page_content
```

```python
# Retriever Option
retriever = vectordb.as_retriever()
retriever.invoke(query)[0].page_content
```
